```python
import numpy as np 
import matplotlib.pyplot as plt
```

creating the time domain function

```python
sr = 1000
T = 1 / sr
N = 1000
t = np.arange(0, N*T , T)
f = np.sin(t) + ((1/2) *np.sin(2*t)) + ((1/3)* np.sin(3*t))
```

creating the noise funciton

```python
noise = np.random.normal(0, 1, len(t))
```

adding the noise

```python
noisy_signal = f + noise
```

turning the time domain function into frequency domain

```python
clean_fourier_signal = np.fft.fft(f)
```

```python
fourier_signal = np.fft.fft(noisy_signal)
```

calculating the frequencies

```python
freq =  np.fft.fftfreq(len(t), d=T)
```

plotting the real and imaginary parts against the frequency

```python
plt.plot(freq, fourier_signal.imag, freq, fourier_signal.real)
plt.show()
```

time domain plot 

```python
plt.figure(figsize=(10, 4))
plt.plot(t, f, label='Clean Signal', linewidth=2)
plt.plot(t, noisy_signal, label='Noisy Signal', alpha=0.6, linestyle='--')
plt.title('Time-Domain Signal')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```

frequency domain plot

```python
plt.figure(figsize=(10, 4))
plt.plot(freq, np.abs(fourier_signal))
plt.title('Frequency-Domain Signal (Magnitude Spectrum)')
plt.xlabel('Frequency (Hz)')
plt.ylabel('|FFT|')
plt.grid(True)
plt.show()
```
