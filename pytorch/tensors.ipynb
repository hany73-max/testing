## GPU usage 

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

## tensors

```python
scalar = torch.tensor(7)
print(scalar)
print(scalar.ndim)
print(scalar.shape)
```

```python
vector = torch.tensor([7, 4])
print(vector)
print(vector.ndim)
print(vector.shape)
```

```python
matrix = torch.tensor([[1,2],
                    [3,4]])
print(matrix)
print(matrix.ndim)
print(matrix.shape)
```

## tensors creation

```python
random_tensor = torch.rand(size=(3, 3, 3))
print(random_tensor)
print(random_tensor.ndim)
print(random_tensor.shape)
```

```python
zeros = torch.zeros(size=(3,3,3))
zeros
```

```python
ones = torch.ones(size=(3,3,3))
ones
```

```python
# creating a tensor u dont know the shape of to match another tensor
# tensor_like = torch.ones_like(random_tensor)
# tensor_like = torch.zeros_like(random_tensor)
tensor_like = torch.rand_like(random_tensor)
tensor_like
```

## tensor manipulation

```python
torch.arange(1,11,1)
```

```python
torch.mm(torch.rand(size=(3,2)), torch.rand(size=(3,2)).T)
```

```python
torch.rand(size= (3,2)).T.shape
```

```python
torch.mean(torch.arange(0,100,10, dtype=torch.float))
```
