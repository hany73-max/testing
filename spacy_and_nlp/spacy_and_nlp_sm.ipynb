```python
import spacy 
from spacy import displacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
with open("../notebooks/text.txt") as f:
    text = f.read()
```

```python
doc = nlp(text)
```

```python
for token in doc[0:10]:
    print(token)
```

```python
for token in  list(doc.sents)[0]:
    print(token)
```

```python
sent =  list(doc.sents)[0]
sent
```

```python
token = sent[2]
token
```

linguistic annotations (attributes)

```python
print(token.right_edge)    # right most component in the subtree of the sentence
print(token.left_edge)     # left most component in the subtree of the sentence
print(token.dep_)          # returns the string representation of the dependency label 
print(token.text)          # the token as a text lol
print([x.text for x in list(token.subtree)])        # output the whole subtree of a token
print([x.text for x in list(token.children)])       # output all the children associated with the specific token
print([x.text for x in list(token.ancestors)])      # all the ancestors of the token
print(token.head)                                   # the most direct ancestor of the token
print(token.ent_type_)                              # the type of the token in this case geo-political entity
print(token.ent_iob_)                               # expresses the state of the token (iob) (inside, outside, beginning) of an entity
print(token.pos_)                                   # provides a universal part-of-speech label like NOUN, VERB, ADJ, or ADV.
print(token.lang_)                                  # returns the language the token written in 
displacy.render(sent, style="dep", jupyter=True)
```

```python
print(sent[12])   
print(sent[12].lemma_)   # returns the origin of the word
```

named entity recognition (NER)

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
displacy.render(doc, style="ent", jupyter=True)
```
